```python
props = ('event', 'name', 'date','drivers','results','laps','car_data',
'pos_data',
'weather_data',
'session_status',
'track_status',
'race_control_messages',
'session_info',
'total_laps',
't0_date',
'session_start_time')

def check_and_create_path(path):
    if not os.path.isdir(path):
        os.mkdir(path)
def get_dataframes(sesh, props, path):
    for prop in props:
        value = getattr(sesh, prop)
        if not os.path.isdir(path):
            os.mkdir(path)
        file_path = path+'/'+str(prop)+'.csv'
        if os.path.isfile(file_path):
            continue
        try:
            #write_dataframe(value, file_path, prop)
            pd.DataFrame(value).to_csv(file_path)
        except Exception as e:
            print(e)

def write_dataframe(value, path, prop):
    #if not os.path.isdir(path):
  #      os.mkdir(path)
   # file_path = path+'/'+str(prop)+'.csv'
   # if os.path.isfile(file_path):
   #     return
    #df = pd.DataFrame(value)
    value.to_csv(path)

def check_if_exists(file_path):
    ret = []
    for prop in props:
        file_path = file_path+'/'+str(prop)+'.csv'

        if not os.path.isfile(file_path):
            ret.append(file_path)

    return ret
    
```

```python
rel_path_base = '../data/data_raw'
for type_of_race in ['Q']: # 'R'
    for year in range(2018, 2027): #22 good
        year_path = rel_path_base + '/' + str(year)
        check_and_create_path(year_path)
        for i in range(1,100):
            try:
                sesh_path = year_path + '/' + f'session_{i}_{type_of_race}'
                check_and_create_path(sesh_path)
                
                
                ret = check_if_exists(sesh_path)
                print(f'Year {year} session {i}')
            except Exception as e :
                print(f'''Error {e} for \n
                Session {i}, Year {year}\n''')
                break
                

print(ret)
```
