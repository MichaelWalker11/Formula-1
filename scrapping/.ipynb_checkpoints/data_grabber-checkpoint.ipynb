```python
props = ('event', 'name', 'date','drivers','results','laps','car_data',
'pos_data',
'weather_data',
'session_status',
'track_status',
'race_control_messages',
'session_info',
'total_laps',
't0_date',
'session_start_time')

def get_dataframes(sesh, props, path):
    for prop in props:
        value = getattr(sesh, prop)
        
        try:
            write_dataframe(value, path, prop)
        except Exception as e:
            print(e)

def write_dataframe(value, path, prop):
    if not os.path.isdir(path):
        os.mkdir(path)
    file_path = path+'/'+str(prop)+'.csv'
    df = pd.DataFrame(value)
    df.to_csv(file_path)


    
```
